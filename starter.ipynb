```python
# Implement baseline algorithm

"""
The Problem: The Maximum Subarray Problem. Given a list of integers (containing both positive and negative numbers), find the contiguous sublist with the largest sum. For example:

Input: [-2, 1, -3, 4, -1, 2, 1, -5, 4]
Output: 6
Explanation: The contiguous sublist [4, -1, 2, 1] has the largest sum = 6.
"""

input_list = [-2, 1, -3, 4, -1, 2, 1, -5, 4]

max_sum = 0

for i in range(len(input_list)):
    current_sum = 0
    for num in range(i, len(input_list)):
        current_sum += input_list[num]
        if current_sum > max_sum:
            max_sum = current_sum
print(max_sum)
print(len(input_list))


```
